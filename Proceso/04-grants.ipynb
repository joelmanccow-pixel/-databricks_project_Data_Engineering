# Permisos sobre objetos del Unity Catalog

```python
CATALOG       = "adbprojectfinaljm"
SCHEMA_BRONZE = "bronze"
SCHEMA_SILVER = "silver"
SCHEMA_GOLDEN = "golden"
```

```python
# 1. PERMISOS A NIVEL DE CATALOG
spark.sql(f"""
    GRANT USE CATALOG ON CATALOG {CATALOG} 
    TO `account users`
""")
```

```python
# 2. PERMISOS A NIVEL DE SCHEMA

# Bronze: solo lectura
spark.sql(f"GRANT USE SCHEMA ON SCHEMA {CATALOG}.{SCHEMA_BRONZE} TO `account users`")
spark.sql(f"GRANT SELECT ON SCHEMA {CATALOG}.{SCHEMA_BRONZE} TO `account users`")
print("Bronze: USE SCHEMA + SELECT granted")
```

```python
# Silver: lectura
spark.sql(f"GRANT USE SCHEMA ON SCHEMA {CATALOG}.{SCHEMA_SILVER} TO `account users`")
spark.sql(f"GRANT SELECT ON SCHEMA {CATALOG}.{SCHEMA_SILVER} TO `account users`")
print(" Silver: USE SCHEMA + SELECT granted")
```

```python
# Golden: lectura
spark.sql(f"GRANT USE SCHEMA ON SCHEMA {CATALOG}.{SCHEMA_GOLDEN} TO `account users`")
spark.sql(f"GRANT SELECT ON SCHEMA {CATALOG}.{SCHEMA_GOLDEN} TO `account users`")
print("Golden: USE SCHEMA + SELECT granted")
```

```python
# 3. PERMISOS A NIVEL DE TABLA - BRONZE
```

```python
for tabla in ["equipos", "avisos", "ordenes"]:
    spark.sql(f"""
        GRANT SELECT ON TABLE {CATALOG}.{SCHEMA_BRONZE}.{tabla}
        TO `account users`
    """)
    print(f"SELECT granted en bronze.{tabla}")
```

```python
# 4. PERMISOS A NIVEL DE TABLA - SILVER
```

```python
for tabla in ["equipos", "avisos", "ordenes"]:
    spark.sql(f"""
        GRANT SELECT ON TABLE {CATALOG}.{SCHEMA_SILVER}.{tabla}
        TO `account users`
    """)
    print(f"SELECT granted en silver.{tabla}")
```

```python
# 5. PERMISOS A NIVEL DE TABLA - GOLDEN
```

```python
for tabla in [
    "kpi_mantenimiento_por_centro",
    "kpi_avisos_por_tipo_prioridad",
    "top_equipos_mayor_mantenimiento",
    "tendencia_mensual_mantenimiento"
]:
    spark.sql(f"""
        GRANT SELECT ON TABLE {CATALOG}.{SCHEMA_GOLDEN}.{tabla}
        TO `account users`
    """)
    print(f"SELECT granted en gold.{tabla}")
```

```python
print("\n📋 Permisos en CATALOG:")
display(spark.sql(f"SHOW GRANTS ON CATALOG {CATALOG}"))

print("\n📋 Permisos en GOLD:")
display(spark.sql(f"SHOW GRANTS ON SCHEMA {CATALOG}.{SCHEMA_GOLDEN}"))
```

```python
# Crear grupos
spark.sql("CREATE GROUP IF NOT EXISTS `GRP_ETL_READER`")
spark.sql("CREATE GROUP IF NOT EXISTS `GRP_ETL_ADMIN`")

# Grants a grupos
spark.sql(f"GRANT USE CATALOG ON CATALOG adbprojectfinaljm TO `GRP_ETL_READER`")
spark.sql(f"GRANT USE SCHEMA ON SCHEMA adbprojectfinaljm.gold TO `GRP_ETL_READER`")
spark.sql(f"GRANT SELECT ON SCHEMA adbprojectfinaljm.gold TO `GRP_ETL_READER`")
print("✅ Grupos y grants aplicados")
```
